```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('airquality.csv')

# Corrected the location filter from 'city_center' to 'city' based on the dataset values
city_data = df[df['Location'] == 'city']

# Calculate mean only for numeric column 'PM2.5_Level'
avg_pollution = df.groupby('Location')[['PM2.5_Level']].mean()
print("Average Pollution by Location:\n", avg_pollution)

# Plot the trend for 'city'
plt.figure(figsize=(8, 4))
plt.plot(city_data['Date'], city_data['Air_Quality_category'], marker='o', color='red')
plt.title('Chart 1: Air_Quality_Category Trend at City Location')
plt.xlabel('Date')
plt.ylabel('Air Quality Category')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Bar Chart (Comparing average PM 2.5 by location)
plt.figure(figsize=(8, 4))
avg_pollution['PM2.5_Level'].plot(kind='bar', color='orange')
plt.title('Chart 2: Average PM2.5 Particles by Location')
plt.xlabel('Location')
plt.ylabel('Average PM2.5')
plt.xticks(rotation=0)
plt.show()
```

```python
# Chart 3: Scatter Plot (Distribution of PM2.5 Levels over Dates)
plt.figure(figsize=(8, 4))
plt.scatter(df['Date'], df['PM2.5_Level'], color='purple', s=50)
plt.title('Chart 3: PM2.5 Level Distribution by Date')
plt.xlabel('Date')
plt.ylabel('PM2.5_Level')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#char 4
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/airquality.csv')

plt.figure(figsize=(8, 4))
plt.hist(df['PM2.5_Level'], bins=20, color='green', edgecolor='black')
plt.title('char 4 :Distribution of PM2.5 Levels')
plt.xlabel('PM2.5 Level')
plt.ylabel('Frequency')
plt.show()
```

```python
#char 5:Horizontal
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].mean().plot(kind='bar', color='blue')
plt.title('char 5 :Average PM2.5 Levels by Location')
plt.xlabel('Peak PM2.5_Level')
plt.ylabel('Location')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
df.groupby('Location')['PM2.5_Level'].mean().plot(kind='barh', color='blue')
plt.title('Chart: Average PM2.5 Levels by Location')
plt.xlabel('Average PM2.5 Level')
plt.ylabel('Location')
plt.show()
```
